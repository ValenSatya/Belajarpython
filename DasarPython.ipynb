## Function, return, argument

```python
# dengan argumen atau parameter
def fungsi_penjumlahan (num1,num2):
    print(num1 + num2)

fungsi_penjumlahan(22,89)
```

```python
# Return
def luas_segitiga(alas,tinggi):
    luas = 1/2 * alas * tinggi
    return luas

print(luas_segitiga(40,97))


```

```python
def perkalian (angka1, angka2):
    return angka1 * angka2

input = perkalian(800,17)
print(input)
```

```python
# default argumen atau parameter

def login (password, username = "user"):
    print(f"selamat datang {username}")

login(123456,"mabar")
login(1234455)
```

```python
# contoh function dengan return dan argument

def mencari_luas_persegi_panjang(panjang,lebar):
    luas_persegi_panjang = panjang * lebar
    return luas_persegi_panjang

persegi_panjang_pertama = mencari_luas_persegi_panjang(50,78)
print(persegi_panjang_pertama)
```

```python
# menggunakan lambda

rumus_luas_persegi = lambda sisi : sisi**2
print(rumus_luas_persegi(8))
```

## Variabel Lokal dan Global

```python
# variabel global

x = 67

def tambah(y):
    hasil = x+y
    return hasil

result = tambah(500)
print(f"hasil dari penjumlahan adalah : {result}")
```

```python
# variabel lokal

def tambah(x, y):
    variabel = 67
    hasil = x+y * variabel
    return hasil


result = tambah(500,90)
print(f"hasil dari penjumlahan adalah : {result}")
```
