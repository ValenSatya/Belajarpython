```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
timeslipping = {'penguasaan': [6,10,14,28,40,57],
                'kesuksesan': [20,40,55.8,69,85,96]}

timeslip_df = pd.DataFrame(timeslipping)
timeslip_df
```

```python
X = np.array(timeslip_df['penguasaan'])
y = np.array(timeslip_df['kesuksesan'])

print(f'X:{X}')
print(f'y:{y}')
```

```python
rata_dev_X = X - np.mean(X)
rata_dev_y = y - np.mean(y)

slope = np.sum(rata_dev_X * rata_dev_y) / np.sum(rata_dev_X**2)
intercept = np.mean(y) - slope * np.mean(X)

print(f'slope = {slope} intercept = {intercept}')
```

```python
y_pred = intercept + slope * X
print(y_pred) 
```

```python
cost_function = np.mean((y_pred - y)**2)
print(cost_function)
```

```python

plt.scatter(X, y, color='blue', label='Data Asli')
plt.plot(X, y_pred, color='red', label='Model OLS')
plt.legend()
plt.show()
```

```python
mae = np.mean(np.abs(y_pred - y))
print(f"MAE: {mae}")
```
