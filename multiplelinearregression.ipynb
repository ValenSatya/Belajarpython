```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sklearn
```

Dataset Training

```python
timeslipping = {'penguasaan': [6,10,14,28,40],
                'stabilitas_emosi': [2, 4, 5, 7, 9],
                'kesuksesan': [20,40,55.8,69,85]}

train_timeslip_df = pd.DataFrame(timeslipping)
train_timeslip_df
```

Dataset Testing

```python
timeslipping = {'penguasaan': [21,32,43,65,76],
                'stabilitas_emosi': [2, 4, 5, 7, 9],
                'kesuksesan': [19,34,42,56.9,87]}

test_timeslip_df = pd.DataFrame(timeslipping)
test_timeslip_df
```

Preprocessing

```python
X_train = np.array(train_timeslip_df[['penguasaan', 'stabilitas_emosi']])
y_train = np.array(train_timeslip_df['kesuksesan'])

print(f'data latih x adalah {X_train}')
print(f'data latih y adalah {y_train}')
```

```python
X_test = np.array(test_timeslip_df[['penguasaan', 'stabilitas_emosi']])
y_test = np.array(test_timeslip_df['kesuksesan'])

print(f'data latih x adalah {X_test}')
print(f'data latih y adalah {y_test}')
```

Modeling

](attachment:image.png)

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(f'r_squared = {r2_score(y_test, y_pred)}')
```
